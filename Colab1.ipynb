# 🚀 Japan-Paw Catalog Indexer en Google Colab (Concurrencia + Rotación de Proxies)

Este notebook ejecuta la indexación masiva en la nube de Google Colab a máxima velocidad:
- **Rotación de Proxies:** Descarga proxies HTTPS/Elite públicos y rota de IP en cada consulta para evitar rate-limits de AnimeTosho y Craftervault.
- **Concurrencia Multi-hilo:** Procesa múltiples episodios en paralelo (8-12 trabajadores simultáneos).

- **Reanudación Segura:** Salta de inmediato los episodios que ya estén verificados en `verified-matches.json`.
