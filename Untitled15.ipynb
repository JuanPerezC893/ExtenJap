```python
import json
import requests
from concurrent.futures import ThreadPoolExecutor, as_completed
import time

# --- Configuración ---
INPUT_FILE = "chapters.json"
OUTPUT_FILE = "reporte_enlaces.json"
MAX_WORKERS = 100  # Número de hilos simultáneos
TIMEOUT = 5       # Segundos para esperar respuesta
# --------------------

def check_link(chapter_data):
    """
    Verifica un solo enlace.
    Retorna una tupla: (estado_booleano, código_estado, datos_capitulo)
    """
    url = chapter_data.get('chapter_url')

    if not url:
        return (False, "No URL", chapter_data)

    try:
        # Intentamos HEAD primero (más rápido)
        response = requests.head(url, timeout=TIMEOUT, allow_redirects=True)

        # Si el servidor no permite HEAD (405), intentamos un GET ligero
        if response.status_code == 405:
            response = requests.get(url, stream=True, timeout=TIMEOUT)
            response.close() # Cerramos conexión inmediatamente

        is_ok = response.status_code == 200
        return (is_ok, response.status_code, chapter_data)

    except requests.RequestException:
        return (False, "Error", chapter_data)

def main():
    print(f"Cargando {INPUT_FILE}...")
    try:
        with open(INPUT_FILE, 'r', encoding='utf-8') as f:
            data = json.load(f)
    except FileNotFoundError:
        print("No se encontró el archivo chapters.json")
        return

    # Aplanar la lista
    all_tasks = []

    for page in data:
        # Aquí capturamos el título de la página (Anime)
        anime_title = page.get('page_title', 'Sin Título')

        for version in page.get('versions', []):
            version_name = version.get('version_name', 'General')
            for chapter in version.get('chapters', []):
                task_data = chapter.copy()
                # Lo guardamos en el diccionario de la tarea para usarlo al imprimir
                task_data['context_anime'] = anime_title
                task_data['context_version'] = version_name
                all_tasks.append(task_data)

    total_links = len(all_tasks)
    print(f"Verificando {total_links} enlaces con {MAX_WORKERS} hilos...")

    start_time = time.time()
    results = {
        "working": [],
        "broken": []
    }

    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
        future_to_link = {executor.submit(check_link, task): task for task in all_tasks}

        completed = 0
        for future in as_completed(future_to_link):
            is_ok, status, task_info = future.result()
            completed += 1

            # Recuperamos el título que guardamos antes
            anime_name = task_info['context_anime']
            chapter_name = task_info['chapter_title']

            # Cortamos el nombre si es muy largo para que la consola se vea ordenada
            display_name = (anime_name[:25] + '..') if len(anime_name) > 25 else anime_name

            item_summary = {
                "anime": anime_name,
                "chapter": chapter_name,
                "url": task_info['chapter_url'],
                "status": status
            }

            if is_ok:
                results["working"].append(item_summary)
                # AQUÍ ESTÁ EL CAMBIO EN EL PRINT:
                print(f"[{completed}/{total_links}] ✅ {status} | [{display_name}] {chapter_name}")
            else:
                results["broken"].append(item_summary)
                # AQUÍ TAMBIÉN:
                print(f"[{completed}/{total_links}] ❌ {status} | [{display_name}] {chapter_name}")

    duration = time.time() - start_time

    print("\n" + "="*40)
    print(f"RESUMEN FINAL ({duration:.2f} segundos)")
    print(f"✅ Funcionando: {len(results['working'])}")
    print(f"❌ Rotos:       {len(results['broken'])}")
    print("="*40)

    if results['broken']:
        with open("enlaces_rotos.json", 'w', encoding='utf-8') as f:
            json.dump(results['broken'], f, indent=2, ensure_ascii=False)
        print("⚠️ Se ha generado 'enlaces_rotos.json' con los detalles.")

if __name__ == "__main__":
    main()
```

```python
import json
import cloudscraper
import logging
from requests.adapters import HTTPAdapter
from concurrent.futures import ThreadPoolExecutor, as_completed
import time

# --- 1. SILENCIAR ALERTAS MOLESTAS ---
# Esto oculta los warnings de "Connection pool is full" para mantener la consola limpia
logging.getLogger("urllib3").setLevel(logging.ERROR)

# --- Configuración ---
INPUT_FILE = "chapters.json"
OUTPUT_FILE = "reporte_enlaces.json"
MAX_WORKERS = 50  # 50 hilos simultáneos
TIMEOUT = 2       # 3 segundos máximo de espera
# --------------------

def check_link(scraper, chapter_data):
    """
    Función optimizada para verificar un enlace usando la sesión compartida.
    """
    url = chapter_data.get('chapter_url')

    if not url:
        return (False, "No URL", chapter_data)

    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
            "Referer": "https://paste.japan-paw.net/"
        }

        # Intentamos HEAD (lo más rápido)
        response = scraper.head(url, timeout=TIMEOUT, headers=headers, allow_redirects=True)

        # Si falla con 405 (Method Not Allowed), 500 o 403, intentamos GET rápido
        if response.status_code in [405, 500, 403]:
            response = scraper.get(url, stream=True, timeout=TIMEOUT, headers=headers)
            response.close() # Cortamos la descarga inmediatamente

        # Consideramos válido 200 (OK) y 206 (Partial Content)
        is_ok = response.status_code in [200, 206]

        return (is_ok, response.status_code, chapter_data)

    except Exception:
        # Cualquier error de conexión (timeout, dns, etc) cuenta como fallo
        return (False, "Error", chapter_data)

def main():
    print(f"Cargando {INPUT_FILE}...")

    # --- 2. CONFIGURACIÓN DEL SCRAPER (Dentro de main para orden) ---
    print("Iniciando motor de scraping y configurando conexiones...")
    scraper = cloudscraper.create_scraper(browser={'browser': 'chrome', 'platform': 'windows', 'mobile': False})

    # Aumentamos la piscina de conexiones para soportar los 50 hilos sin errores
    adapter = HTTPAdapter(pool_connections=MAX_WORKERS, pool_maxsize=MAX_WORKERS)
    scraper.mount("https://", adapter)
    scraper.mount("http://", adapter)

    try:
        with open(INPUT_FILE, 'r', encoding='utf-8') as f:
            data = json.load(f)
    except FileNotFoundError:
        print("❌ No se encontró el archivo chapters.json")
        return

    # Aplanar la lista de enlaces
    all_tasks = []
    for page in data:
        anime_title = page.get('page_title', 'Sin Título')
        for version in page.get('versions', []):
            version_name = version.get('version_name', 'General')
            for chapter in version.get('chapters', []):
                task_data = chapter.copy()
                task_data['context_anime'] = anime_title
                task_data['context_version'] = version_name
                all_tasks.append(task_data)

    total_links = len(all_tasks)
    print(f"🚀 Verificando {total_links} enlaces con {MAX_WORKERS} hilos...")

    start_time = time.time()
    results = {"working": [], "broken": []}

    # Ejecución Multihilo
    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
        # Pasamos el objeto 'scraper' a cada hilo
        future_to_link = {executor.submit(check_link, scraper, task): task for task in all_tasks}

        completed = 0
        for future in as_completed(future_to_link):
            is_ok, status, task_info = future.result()
            completed += 1

            anime_name = task_info['context_anime']
            chapter_name = task_info['chapter_title']
            display_name = (anime_name[:25] + '..') if len(anime_name) > 25 else anime_name

            icon = "✅" if is_ok else "❌"

            # Imprimimos el progreso
            print(f"[{completed}/{total_links}] {icon} {status} | [{display_name}] {chapter_name}")

            item_summary = {
                "anime": anime_name,
                "chapter": chapter_name,
                "url": task_info['chapter_url'],
                "status": str(status)
            }

            if is_ok:
                results["working"].append(item_summary)
            else:
                results["broken"].append(item_summary)

    duration = time.time() - start_time

    print("\n" + "="*40)
    print(f"🏁 RESUMEN FINAL ({duration:.2f} segundos)")
    print(f"✅ Funcionando: {len(results['working'])}")
    print(f"❌ Rotos:       {len(results['broken'])}")
    print("="*40)

    if results['broken']:
        with open("enlaces_rotos.json", 'w', encoding='utf-8') as f:
            json.dump(results['broken'], f, indent=2, ensure_ascii=False)
        print("⚠️ Se generó 'enlaces_rotos.json' con los detalles.")

if __name__ == "__main__":
    main()
```

```python
import json
import cloudscraper
import logging
import random
import time
from requests.adapters import HTTPAdapter
from fp.fp import FreeProxy  # Librería para buscar proxies gratis
from concurrent.futures import ThreadPoolExecutor, as_completed

# --- CONFIGURACIÓN ---
INPUT_FILE = "chapters.json"
MAX_WORKERS = 50      # Mantenlo bajo para proxies gratuitos (son lentos)
TIMEOUT = 10
# --------------------

logging.getLogger("urllib3").setLevel(logging.ERROR)

def get_proxy_scraper():
    """
    Crea un scraper que usa un proxy aleatorio nuevo.
    """
    try:
        # Busca un proxy fresco (HTTPS y Elite para anonimato)
        proxy = FreeProxy(https=True, elite=True, timeout=1).get()
        proxies = {"http": proxy, "https": proxy}
    except Exception:
        proxies = None # Fallback sin proxy si falla

    scraper = cloudscraper.create_scraper(browser={'browser': 'chrome', 'platform': 'windows', 'mobile': False})

    # Montamos adaptadores
    adapter = HTTPAdapter(max_retries=2)
    scraper.mount("https://", adapter)
    scraper.mount("http://", adapter)

    if proxies:
        scraper.proxies.update(proxies)

    return scraper, proxies

def check_link(chapter_data):
    url = chapter_data.get('chapter_url')
    if not url: return ("NO_URL", chapter_data)

    # Reintentos con diferentes proxies si falla
    for attempt in range(3):
        scraper, current_proxy = get_proxy_scraper()

        try:
            # Pausa humana aleatoria
            time.sleep(random.uniform(1, 3))

            headers = {
                "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
                "Referer": "https://paste.japan-paw.net/"
            }

            # Usamos GET porque los proxies gratuitos a veces fallan con HEAD
            response = scraper.get(url, stream=True, timeout=TIMEOUT, headers=headers)
            status = response.status_code
            response.close()

            if status in [200, 206]:
                return ("OK", chapter_data)
            elif status == 404:
                return ("ROTO", chapter_data)
            elif status == 403:
                # Si da 403, es probable culpa del proxy o IP, reintentamos con otro
                continue

        except Exception:
            continue # Error de conexión del proxy, probamos otro

    # Si falla tras 3 intentos con proxies distintos
    return ("FALLO_RED", chapter_data)

def main():
    print("🚀 Iniciando modo EVASIÓN con Proxies Rotativos...")
    print("⚠️ Nota: Esto será más lento porque busca proxies nuevos, pero evitará el bloqueo.")

    try:
        with open(INPUT_FILE, 'r', encoding='utf-8') as f:
            data = json.load(f)
    except FileNotFoundError:
        print("❌ No se encontró chapters.json")
        return

    all_tasks = []
    for page in data:
        for version in page.get('versions', []):
            for chapter in version.get('chapters', []):
                task = chapter.copy()
                task['anime'] = page.get('page_title', 'Unknown')
                all_tasks.append(task)

    total = len(all_tasks)
    stats = {"OK": 0, "ROTO": 0, "FALLO": 0}

    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
        future_to_link = {executor.submit(check_link, task): task for task in all_tasks}

        completed = 0
        for future in as_completed(future_to_link):
            status, info = future.result()
            completed += 1

            icon = "✅" if status == "OK" else "❌"
            if status == "FALLO_RED": icon = "⚠️"

            print(f"[{completed}/{total}] {icon} {status} | {info['anime'][:20]}... - {info['chapter_title']}")

            if status == "OK": stats["OK"] += 1
            elif status == "ROTO": stats["ROTO"] += 1
            else: stats["FALLO"] += 1

    print(f"\nResumen: ✅ {stats['OK']} | ❌ {stats['ROTO']} | ⚠️ {stats['FALLO']}")

if __name__ == "__main__":
    main()
```
